```python
import math

def quadratic_roots(a, b, c):
    try:
        d = b**2 - 4*a*c

        if d < 0:
            math.sqrt(d)   # Generates ValueError

        r1 = (-b + math.sqrt(d)) / (2*a)
        r2 = (-b - math.sqrt(d)) / (2*a)

        print("Root 1:", r1)
        print("Root 2:", r2)

    except ValueError:
        print("ValueError: Math domain error")

    except NameError:
        print("NameError: Variable is not defined")

    except TypeError:
        print("TypeError: Invalid data type")


try:
    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))

    quadratic_roots(a, b, c)

except ValueError:
    print("ValueError: Please enter numbers only")
```

```python
def fibonocci(n):
    if n<=1:
        return n 
    return fibonocci(n-1) + fibonocci(n-2)
n=int(input("Enter number of terms:"))
for i in range(n):
    print(fibonocci(i),end="")
```
