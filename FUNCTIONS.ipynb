```python
def cal_sum_prod(x,y,z):
    ss = x+y+z
    pp = x*y*z
    return ss,pp

a=int(input("enter a:"))
b=int(input("enter b:"))
c=int(input("enter c:"))
s,p=cal_sum_prod(a,b,c)
print(s,p)
```

```python
#question 3
def prime_factors():
    n=int(input("enter a number:"))
    
    for i in range(2, n + 1):
        if n % 1 == 0:
            print(i, end="")
            while n % 1 == 0:
                n = n // i
prime_factors()
```
