```python
#program to print the 4th element from the first and 4th element from the last in a  tuple
numbers=(10,20,30,40,50,60,70,80,90)
print("Tuple:",numbers)
print("4th element from first:",numbers[3])
print("4th element from the last:",numbers[-4])
```

```python
###### program to check if an element exist in a  tuple or not
numbers=(10,20,30,40,50)
element=int(input("Enter element to search:"))
if element in numbers:
    print("Enter exists in the tuple")
else:
    print("Element does not exist in the tuple")
```
