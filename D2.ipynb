```python
#dictionary
my_dict={
    "name":"ABC",
    "panel":"B",
    "roll no":34,
    "marks":[65,87,97,94]
}
sorted_dict=sorted(my_dict)
print("sorted keys:",sorted_dict)
```

```python
#dictionary with highest marks
students = {
    "Aarav": 85,
    "Diya": 92,
    "Kabir": 78,
    "Ananya": 90,
    "Rohan": 88
}

highest_student=max(students,key=students.get)
print("Student with highest marks:",highest_student)
print("Highest Marks:",students[highest_student])
```

```python
#To count the frequency of each element in a list
numbers = [10,20,30,40,50,10,20,30]
frequency={}
for num in numbers:
    if num in frequency:
        frequency[num]=frequency[num]+1
    else:
        frequency[num]=1
print("Frequency of elements:",frequency)
```

```python
my_dict={
    "Aarav": 85,
    "Diya": 92,
    "Kabir": 78,
    "Ananya": 90,
    "Rohan": 88
}
sorted_dict=dict(sorted(my_dict.items(),key=lambda item:item[1]))
print("Dictionary sorted by values:",sorted_dict)
```
