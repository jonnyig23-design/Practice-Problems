```python
my_dict={
101:"Amit",
102:"Rahul",
103:"Priya"
}
print("Keys:")
for key in my_dict:
    print(key)
    print("\nValues:")

for value in my_dict.values():
    print(value)
print("Keys and Values:")
for key,value in my_dict.items():
    print(key,value)
```
